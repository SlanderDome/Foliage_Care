```python
import tensorflow as tf
from tensorflow.keras.preprocessing.image import ImageDataGenerator
from tensorflow.keras.applications import MobileNetV2
from tensorflow.keras import layers, models
import os
from google.colab import files # For downloading

# Path to your dataset
train_dir = "/content/plant detect"

# Create a binary dataset
datagen = ImageDataGenerator(
    rescale=1.0/255,
    validation_split=0.2
)

print("Loading Training Data:")
train_data = datagen.flow_from_directory(
    train_dir,
    target_size=(224, 224),
    batch_size=32,
    class_mode='binary',
    subset='training'
)

print("Loading Validation Data:")
val_data = datagen.flow_from_directory(
    train_dir,
    target_size=(224, 224),
    batch_size=32,
    class_mode='binary',
    subset='validation'
)

# Load pre-trained MobileNetV2
base_model = MobileNetV2(weights='imagenet', include_top=False, input_shape=(224,224,3))
base_model.trainable = False

# Build model
model = models.Sequential([
    base_model,
    layers.GlobalAveragePooling2D(),
    layers.Dense(64, activation='relu'),
    layers.Dropout(0.3),
    layers.Dense(1, activation='sigmoid')
])

model.compile(
    optimizer='adam',
    loss='binary_crossentropy',
    metrics=['accuracy']
)

print("Starting model training (on GPU)...")

# Train
history = model.fit(
    train_data,
    validation_data=val_data,
    epochs=15
)

# Save and download the model
model.save("new_plant_detector.h5")
print("✅ Model saved as new_plant_detector.h5")

print("Downloading your model...")
files.download("new_plant_detector.h5")
```
